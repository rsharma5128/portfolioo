<h1>SASS Typography: Solved Runners</h1>
<p>Every runner from the lesson, refactored into semantic HTML with OCS typography classes. No custom CSS, <code>&lt;style&gt;</code> blocks, or inline styles remain.</p>


<h2>Example: OCS Role</h2>
<p>The lesson intro already uses the right structure: an <code>&lt;h2&gt;</code>, an <code>ocs_lead</code> paragraph, and an ordered list.</p>


```python
%%html
<!-- UI_RUNNER: Typography Intro - OCS Role | show_code: true -->
<h2>Cat Care 101</h2>
<p class="ocs_lead">Follow these basic rules to keep your feline friend happy.</p>
<ol>
  <li>Provide fresh water and food daily.</li>
  <li>Give them plenty of cozy nap spots.</li>
</ol>
```

<h2>Example A: Basic Headings and OCS Lead</h2>
<p><code>&lt;h2&gt;</code> and <code>&lt;h3&gt;</code> provide the hierarchy while <code>ocs_section-title</code> and <code>ocs_lead</code> give the text its role.</p>


```python
%%html
<!-- UI_RUNNER: Typography Example A - Headings and Lead | show_code: true -->
<h2 class="ocs_section-title">Meet Whiskers</h2>
<h3 class="ocs_lead">The Sleepiest Cat in the World</h3>
<p class="ocs_text">Whiskers spends 18 hours a day napping in sunbeams.</p>
```

<h2>Example B: Emphasis Inside a Component</h2>
<p>The card keeps the text grouped, while <code>&lt;strong&gt;</code> and <code>&lt;em&gt;</code> carry real meaning instead of visual styling.</p>


```python
%%html
<!-- UI_RUNNER: Typography Example B - Emphasis | show_code: true -->
<div class="ocs_card">
  <p class="ocs_text">Please <strong>do not</strong> rub the belly! It is a trap!!</p>
  <p class="ocs_text">It is <em>highly recommended</em> to scratch behind the ears instead!!</p>
</div>
```

<h2>Example C: Full Section Structure</h2>
<p>A heading, description, and unordered list form a complete section inside one card.</p>


```python
%%html
<!-- UI_RUNNER: Typography Example C - Full Section | show_code: true -->
<div class="ocs_card">
  <h2 class="ocs_section-title">Favorite Cat Toys</h2>
  <p class="ocs_description">Approved toys for maximum zoomies include:</p>
  <ul>
    <li>Feather wands</li>
    <li>Cardboard boxes (obviously)</li>
    <li>Laser pointers</li>
  </ul>
</div>
```

<h2>Popcorn Hack: Refactor the Custom CSS</h2>
<p>The <code>&lt;style&gt;</code> block and the <code>large-title</code>, <code>sub-title</code>, and <code>normal</code> classes are gone. The heading, subheading, and paragraph now use semantic tags plus OCS roles.</p>


```python
%%html
<!-- UI_RUNNER: Typography Popcorn - Semantic Refactor | show_code: true -->
<h2 class="ocs_section-title">Why Cats Purr</h2>
<h3 class="ocs_lead">A Sign of Happiness</h3>
<p class="ocs_text">Cats usually purr when they feel safe, cozy, and ready for a nap.</p>
```

<h2>Homework Hack: How to Pat a Cat</h2>
<p>Refactored into one OCS card with a real heading hierarchy, paragraphs, an ordered list, and <code>&lt;strong&gt;</code> emphasis. The fake numbered paragraphs became a real <code>&lt;ol&gt;</code>.</p>


```python
%%html
<!-- UI_RUNNER: Typography Homework - How to Pat a Cat | show_code: true -->
<div class="ocs_card">
  <h2 class="ocs_section-title">How to Pat a Cat</h2>
  <p class="ocs_lead">Have you ever wanted to...</p>
  <p class="ocs_text">pat a cat? Well good for you we will give step by step instructions on how to pat a cat!</p>
  <p class="ocs_text"><strong>How to pat a cat:</strong></p>
  <ol>
    <li>put out your hand</li>
    <li>touch the cat</li>
    <li>move your hand back and forth</li>
  </ol>
  <h3 class="ocs_section-title">Congrats! You pat a cat :D</h3>
</div>
```
