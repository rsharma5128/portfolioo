```python
<!-- UI_RUNNER: Toggle Homework - Study Dashboard | show_code: true -->
<style>
  /* .panel changes look when the dark toggle is on (JS adds/removes the "dark" class) */
  .hw-panel { padding: 16px; border-radius: 12px; border: 1px solid #6B7280; background: #FFFFFF; color: #111111; max-width: 480px; }
  .hw-panel.dark { background: #1F2937; color: #F9FAFB; }
  .hw-panel label { display: block; margin: 10px 0; }
</style>

<div id="hw-panel" class="hw-panel">
  <h3>My Study Dashboard</h3>

  <!-- Toggle 1: shows or hides a study tip -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-tip-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show Study Tip</span>
  </label>

  <!-- Toggle 2: switches the whole panel between light and dark -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-dark-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Dark Mode</span>
  </label>

  <!-- Toggle 3: changes the notification message -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-notif-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Notifications</span>
  </label>

  <!-- Toggle 4: shows or hides a mascot -->
  <label class="ocs__toggle ocs__toggle--green">
    <input id="hw-mascot-toggle" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Study Buddy</span>
  </label>

  <hr>
  <p id="hw-tip" hidden>💡 Tip: study in 25-minute blocks, then take a 5-minute break.</p>
  <p id="hw-notif-status"></p>
  <p id="hw-mascot" hidden>🦉 Hoot hoot! I'm your study buddy.</p>

  <!-- Live counter of how many toggles are currently ON -->
  <p><strong>Toggles on:</strong> <output id="hw-count" aria-live="polite">0</output> of 4</p>
  <p id="hw-summary"></p>
</div>

<script>
  // Grab every element once, using the ids we gave them.
  const tipToggle = document.getElementById("hw-tip-toggle");
  const darkToggle = document.getElementById("hw-dark-toggle");
  const notifToggle = document.getElementById("hw-notif-toggle");
  const mascotToggle = document.getElementById("hw-mascot-toggle");

  const panel = document.getElementById("hw-panel");
  const tip = document.getElementById("hw-tip");
  const notifStatus = document.getElementById("hw-notif-status");
  const mascot = document.getElementById("hw-mascot");
  const countOutput = document.getElementById("hw-count");
  const summary = document.getElementById("hw-summary");

  const allToggles = [tipToggle, darkToggle, notifToggle, mascotToggle];

  function updateDashboard() {
    // I thought .checked was interesting because it lets JavaScript tell whether
    // the toggle is currently turned on or off (true/false) - it's just a Boolean!
    tip.hidden = !tipToggle.checked;           // hidden is also a Boolean, so ! flips it
    mascot.hidden = !mascotToggle.checked;

    // classList.toggle(name, force) adds the class when force is true and removes it when false
    panel.classList.toggle("dark", darkToggle.checked);

    // A conditional chooses which message to show - the same IF/ELSE idea from AP CSP selection
    if (notifToggle.checked) {
      notifStatus.textContent = "🔔 Notifications are ON: you'll get homework reminders.";
    } else {
      notifStatus.textContent = "🔕 Notifications are OFF.";
    }

    // Count the toggles that are on: loop over the array and add 1 for every checked box
    let on = 0;
    const names = [];
    allToggles.forEach(function (toggle) {
      if (toggle.checked) {
        on++;
        names.push(toggle.id.replace("hw-", "").replace("-toggle", ""));
      }
    });
    countOutput.textContent = on;
    summary.textContent = on === 0 ? "Everything is off." : "On: " + names.join(", ");
  }

  // Question I had: do I need a separate listener per toggle? No - the same function
  // can be attached to every toggle's "change" event, so it recalculates everything each time.
  allToggles.forEach(function (toggle) {
    toggle.addEventListener("change", updateDashboard);
  });

  updateDashboard();   // run once so the page starts in the correct state
</script>
```
