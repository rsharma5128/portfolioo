## Popcorn Hack 2: Size the Beakers with `fr`

Four boxes in one row means **four** `fr` values. The Control Beaker is last and should be twice as wide as each of the others, so the values are `1fr 1fr 1fr 2fr`.

**fr values used:** `1fr 1fr 1fr 2fr`

````python
%%html
<!-- UI_RUNNER: Grids Popcorn 2 -->
<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">
  <div class="ocs__grid-cell">Beaker 1</div>
  <div class="ocs__grid-cell">Beaker 2</div>
  <div class="ocs__grid-cell">Beaker 3</div>
  <div class="ocs__grid-cell">Control Beaker</div>
</div>
````

## MCQ Check

| # | Question | My answer | Why |
| --- | --- | --- | --- |
| 1 | Make a standard grid show 4 boxes per row | **A** (`cols-4`) | Same as `1fr 1fr 1fr 1fr`. |
| 2 | `grid-template-columns: 2fr 1fr` | **B** (2 boxes, first twice as wide) | Two values = two boxes; 2 of 3 shares vs. 1 of 3. |
| 3 | Cell type that stretches across the whole row | **C** (`ocs__grid-cell--header`) | `--header` spans every column; `--wide` only spans 2. |
| 4 | Line that breaks the rule | **D** (`class="my-row" style="display:flex"`) | Made-up class plus an inline style. |

## Homework Hack

Fixed both grids using only OCS classes:
- **Grid 1:** `ocs__grid ocs__grid--standard`. The title is a `--header` cell and Beaker 3 (the outlier at 31.6) is an `--accent` cell.
- **Grid 2:** `ocs__grid ocs__grid--calculator`, a keypad. The `Avg` key is an `--accent` cell.
- Every box is an `ocs__grid-cell`, and there are no `style` attributes and no made-up classes (`my-row`, `title-box`, `sample-card`, `highlight`, `key`, `orange` are all gone).

````python
%%html
<!-- UI_RUNNER: Grids Homework -->
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
  <div class="ocs__grid-cell">Beaker 1: 22.4</div>
  <div class="ocs__grid-cell">Beaker 2: 22.9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
</div>

<div class="ocs__grid ocs__grid--calculator">
  <div class="ocs__grid-cell">7</div>
  <div class="ocs__grid-cell">8</div>
  <div class="ocs__grid-cell">9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
</div>
````
