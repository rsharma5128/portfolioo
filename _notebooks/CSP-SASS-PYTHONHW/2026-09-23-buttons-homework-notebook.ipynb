## Popcorn Hack #1: Add the Base Button Class

Added `ocs__btn` to the empty `class=""` so the link gets OCS button styling.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 1 - Base Button Class -->
<a class="ocs__btn" href="https://www.mygoodbrain.org/resources"> Mental Health Resource Directory</a>
```

## Popcorn Hack #2: Modifiers & Button Groups

1. Added `pill` and `fill` to the "Complete Daily Check-In" button (it keeps `accent`).
2. Replaced the plain `<div>` with `<div class="ocs__links">` so both actions are grouped.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 2 - Modifiers & Button Groups -->
<div class="ocs__links">
  <button class="ocs__btn accent pill fill">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="">Save Draft Journal</a>
</div>
```

## Popcorn Hack #3: Status Tones & Wide Layout Bar

1. Changed the status tone from `alert-yellow` to `alert-green` on "Complete Check-In"
2. Upgraded the container from `ocs__links` to `ocs__links--wide`.
3. Added `fill` to the Crisis Support button for full visual priority.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 3 - Status Tones & Wide Layout -->
<div class="ocs__links--wide">
  <button class="ocs__btn alert-green outline">Complete Check-In</button>
  <a class="ocs__btn alert-red fill" href="https://www.mygoodbrain.org/resources"> Get Crisis Support</a>
</div>
```

## Homework Hack #1: Adaptive Resource Toolbar

All three actions are wrapped in `ocs__links`, every inline `style` and hex/color word is gone, and each category uses the class set for its urgency:
- **Urgent Support:** `ocs__btn alert-red fill pill`
- **Self-Care:** `ocs__btn alert-green fill pill`
- **Community:** `ocs__btn accent outline pill`

```python
%%html
<!-- UI_RUNNER: Buttons Homework 1 - Adaptive Resource Toolbar -->
<div class="ocs__links">
  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources">Urgent Support</a>
  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources">Self-Care</a>
  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources">Community</a>
</div>
```

## Homework Hack #2: Primary vs. Secondary Action Priority

Wrapped everything in `ocs__links`, removed all inline styles (including the outer `<div>`'s background/padding), and built a visual hierarchy:
- **Primary** ("Log Mood Entry"): `ocs__btn accent fill`, the solid, highest-priority action.
- **Secondary** ("View Weekly Trends"): `ocs__btn outline`, medium priority.
- **Alert** ("Crisis Hotline"): `ocs__btn alert-red outline`, red so it stands out as urgent support without competing with the primary action.

```python
%%html
<!-- UI_RUNNER: Buttons Homework 2 - Primary vs. Secondary Action Priority -->
<div class="ocs__links">
  <button class="ocs__btn accent fill">Log Mood Entry</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources">View Weekly Trends</a>
  <a class="ocs__btn alert-red outline" href="https://www.mygoodbrain.org/resources">Crisis Hotline</a>
</div>
```
